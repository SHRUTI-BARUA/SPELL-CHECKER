```python
import re
from collections import Counter, defaultdict

def load_corpus(file_path):

    with open(file_path,'r',encoding='utf8') as f:
        text = f.read().lower()

    words = re.findall(r'\w+', text)

    return words


words = load_corpus(DATA_PATH)

print("Total words:", len(words))
print("Sample:", words[:20])
```

```python
vocab = set(words)

unigram_counts = Counter(words)

total_words = sum(unigram_counts.values())

V = len(vocab)

print("Vocabulary size:", V)
```

```python
def unigram_prob(word):

    return (unigram_counts[word] + 1) / (total_words + V)
```

```python
bigram_counts = defaultdict(int)

for i in range(len(words)-1):

    bigram_counts[(words[i],words[i+1])] += 1
```

```python
def bigram_prob(prev_word,word):

    numerator = bigram_counts[(prev_word,word)] + 1

    denominator = unigram_counts[prev_word] + V

    return numerator / denominator
```

```python
trigram_counts = defaultdict(int)

for i in range(len(words)-2):

    trigram_counts[(words[i],words[i+1],words[i+2])] += 1
def trigram_prob(w1,w2,w3):

    numerator = trigram_counts[(w1,w2,w3)] + 1

    denominator = bigram_counts[(w1,w2)] + V

    return numerator / denominator
```

```python
alphabet = "abcdefghijklmnopqrstuvwxyz"

def edits1(word):

    splits = [(word[:i],word[i:]) for i in range(len(word)+1)]

    deletes = [L + R[1:] for L,R in splits if R]

    transposes = [L + R[1] + R[0] + R[2:]
                  for L,R in splits if len(R)>1]

    replaces = [L + c + R[1:]
                for L,R in splits if R for c in alphabet]

    inserts = [L + c + R
               for L,R in splits for c in alphabet]

    return set(deletes + transposes + replaces + inserts)
```

```python
def known(words):

    return set(w for w in words if w in vocab)


def candidates(word):

    return (
        known([word]) or
        known(edits1(word)) or
        [word]
    )
```

```python
def error_prob(typo,word):

    if typo == word:
        return 1

    if word in edits1(typo):
        return 0.1

    return 0.01
```

```python
def correct_unigram(word):

    return max(
        candidates(word),
        key=lambda w: error_prob(word,w) * unigram_prob(w)
    )
```

```python
def correct_bigram(prev_word,typo):

    return max(
        candidates(typo),
        key=lambda w: error_prob(typo,w) * bigram_prob(prev_word,w)
    )
```

```python
def correct_trigram(w1,w2,typo):

    return max(
        candidates(typo),
        key=lambda w: error_prob(typo,w) * trigram_prob(w1,w2,w)
    )
```

```python
def correct_sentence(sentence):

    tokens = re.findall(r'\w+',sentence.lower())

    corrected = tokens[:2]

    for i in range(2,len(tokens)):

        w1 = corrected[i-2]
        w2 = corrected[i-1]

        corrected_word = correct_trigram(w1,w2,tokens[i])

        corrected.append(corrected_word)

    return " ".join(corrected)
```

```python
test_set = [

("connection","conection"),
("router","rounter"),
("internet","internat"),
("ethernet","ethrnet"),
("signal","singal"),
("disconnecting","disconecting")

]
```

```python
def evaluate(model):

    correct = 0

    total = len(test_set)

    for right, typo in test_set:

        pred = model(typo)

        if pred == right:

            correct += 1

    precision = correct/total
    recall = correct/total

    f1 = 2*(precision*recall)/(precision+recall)

    return precision,recall,f1
```

```python
p,r,f = evaluate(correct_unigram)

print("Unigram Results")
print("Precision:",p)
print("Recall:",r)
print("F1:",f)
```

```python


# Test set: (correct_word , typo)

test_set = [

("connection","conection"),
("router","rounter"),
("internet","internat"),
("ethernet","ethrnet"),
("signal","singal"),
("network","netwrok"),
("driver","drvier"),
("password","passwrod")

]

# Unigram wrapper
def unigram_model(word):
    return correct_unigram(word)


# Bigram wrapper (use a default previous word)
def bigram_model(word):
    return correct_bigram("wifi", word)


# Trigram wrapper (use default context)
def trigram_model(word):
    return correct_trigram("wifi", "signal", word)
def evaluate_model(model):

    correct = 0
    total_errors = len(test_set)
    total_predictions = 0

    for correct_word, typo in test_set:

        prediction = model(typo)

        total_predictions += 1

        if prediction == correct_word:
            correct += 1

    precision = correct / total_predictions
    recall = correct / total_errors

    if precision + recall == 0:
        f1 = 0
    else:
        f1 = 2 * (precision * recall) / (precision + recall)

    return precision, recall, f1
```

```python
# Evaluate Unigram
p1, r1, f1 = evaluate_model(unigram_model)

# Evaluate Bigram
p2, r2, f2 = evaluate_model(bigram_model)

# Evaluate Trigram
p3, r3, f3 = evaluate_model(trigram_model)
```
